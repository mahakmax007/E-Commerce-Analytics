```python
rfm_data = (
    orders
    .merge(
        customers[["customer_id", "customer_unique_id"]],
        on="customer_id",
        how="left"
    )
    .merge(
        order_items[["order_id", "price"]],
        on="order_id",
        how="left"
    )
)

rfm_data.head()
```

```python
rfm = (
    rfm_data[rfm_data["order_status"] == "delivered"]
    .groupby("customer_unique_id")
    .agg(
        Recency=("order_purchase_timestamp", 
                 lambda x: (rfm_data["order_purchase_timestamp"].max() - x.max()).days),
        Frequency=("order_id", "nunique"),
        Monetary=("price", "sum")
    )
)

rfm.head()
```

```python
rfm["R_score"] = pd.qcut(
    rfm["Recency"],
    5,
    labels=[5, 4, 3, 2, 1]
)

rfm["F_score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
)

rfm["M_score"] = pd.qcut(
    rfm["Monetary"],
    5,
    labels=[1, 2, 3, 4, 5]
)

rfm.head()
```

```python
rfm["RFM_Score"] = (
    rfm["R_score"].astype(int).astype(str)
    + rfm["F_score"].astype(int).astype(str)
    + rfm["M_score"].astype(int).astype(str)
)

rfm.head()
```

```python
id="rfm04"
def segment_customer(row):
    if row["RFM_Score"] == "555":
        return "Champions"
    elif row["R_score"] >= 4 and row["F_score"] >= 4:
        return "Loyal Customers"
    elif row["R_score"] >= 4 and row["M_score"] >= 4:
        return "Potential Loyalists"
    elif row["R_score"] <= 2 and row["F_score"] >= 3:
        return "At Risk"
    elif row["R_score"] <= 2 and row["M_score"] >= 3:
        return "Cannot Lose Them"
    elif row["R_score"] <= 2 and row["F_score"] <= 2:
        return "Lost Customers"
    else:
        return "Others"

rfm["Segment"] = rfm.apply(segment_customer, axis=1)

print(rfm["Segment"].value_counts())
```

```python
champions = rfm[rfm["Segment"] == "Champions"]

print("Champions Customers:", len(champions))
print("Champions Revenue:", round(champions["Monetary"].sum(), 2), "BRL")
print("Average Champions Spend:", round(champions["Monetary"].mean(), 2), "BRL")
```

```python
at_risk = rfm[rfm["Segment"] == "At Risk"]

print("At Risk Customers:", len(at_risk))
print("At Risk Revenue:", round(at_risk["Monetary"].sum(), 2), "BRL")
print("Average At Risk Spend:", round(at_risk["Monetary"].mean(), 2), "BRL")
```

```python
lost = rfm[rfm["Segment"] == "Lost Customers"]

print("Lost Customers:", len(lost))
print("Lost Customer Revenue:", round(lost["Monetary"].sum(), 2), "BRL")
print("Average Lost Customer Spend:", round(lost["Monetary"].mean(), 2), "BRL")
```
